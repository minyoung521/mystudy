```python
# 파이썬은 문자 데이터와 문자열 데이터를 구분하지 않고 큰따옴표나 작은따옴표로 짝을 맞춰서 입력하면 된다.
# print() 함수는 별도의 설정을 하지 않으면 ()안의 내용을 출력하고 줄을 바꾼다.
print('Hello Python')
print("안녕 파이썬")
print("제2외국어를 배운 호랑이가 '야옹' 했습니다.")
print("제2외국어를 배운 호랑이가 \"야옹\" 했습니다.")
```

```python
# print() 함수로 출력한 후 줄을 변경하지 않으려면 end 인자를 사용한다.
# end 인자는 print() 함수로 출력한 후 출력할 내용을 지정한다.
# end 인자는 기본값이 '\n' 이고 '\n'은 new line이라는 의미이며 개행 문자를 의미한다.
print('Hello Python', end = '\n')
print("안녕 파이썬")
print('Hello Python', end = '')
print("안녕 파이썬")
print('Hello Python', end = ' ')
print("안녕 파이썬")
print('Hello Python', end = ': ')
print("안녕 파이썬")
```

# 문자열 연산자

문자열과 문자열에 `+` 연산자를 사용하면 문자열 연결

```python
print(5 + 8)
print('5' + '8')
# 아래와 같이 문자열과 숫자 또는 숫자와 문자열 데이터에 + 연산자를 사용하면 데이터 타입이 달라 에러가 발생된다.
# print('5' + 8) # 에러
# print(5 + '8') # 에러
```

```python
# 문자열을 숫자로 변환하려면 int() 또는 float() 함수를 사용한다.
print(int('5') + 8)
# 숫자를 문자열로 변환하려면 str() 함수를 사용한다.
print('5' + str(8))
```

문자열에 `*` 연산자를 사용하면 문자열이 반복된다.

```python
print('안녕하세요' * 100)
```

# 출력 서식 지정하기
format() 메소드 사용

```python
print("{} + {} = {}".format(5, 3, 5+3))
print("{} - {} = {}".format(5, 3, 5-3))
print("{} * {} = {}".format(5, 3, 5*3))
print("{} / {} = {:.2f}".format(5, 3, 5/3))
print("{} // {} = {}".format(5, 3, 5//3))
print("{} ** {} = {}".format(5, 3, 5**3))
```

```python
print(f"{5} + {3} = {5+3}")
print(f"{5} - {3} = {5-3}")
print(f"{5} * {3} = {5*3}")
print(f"{5} / {3} = {5/3:.2f}")
print(f"{5} // {3} = {5//3}")
print(f"{5} % {3} = {5%3}")
print(f"{5} ** {3} = {5**3}")
```

```python
print("{}는(은) {}원 입니다.".format("메론", 15000))
print("{:10s}는(은) {}원 입니다.".format("메론", 15000))
print("{:<10s}는(은) {}원 입니다.".format("메론", 15000))
print("{:>10s}는(은) {}원 입니다.".format("메론", 15000))
print("{:^10s}는(은) {}원 입니다.".format("메론", 15000))
print("{:_<10s}는(은) {}원 입니다.".format("메론", 15000))
print("{:_>10s}는(은) {}원 입니다.".format("메론", 15000))
print("{:_^10s}는(은) {}원 입니다.".format("메론", 15000))
print("{}는(은) {:.2f}원 입니다.".format("메론", 15000))
print("{}는(은) {:7,d}원 입니다.".format("메론", 15000)) #숫자 1000자리 마다 , 찍기
print("{}는(은) {:,.2f}원 입니다.".format("메론", 15000))
print("{}는(은) {:,.2%}원 입니다.".format("메론", 15000))
```

```python
name = "메론"
price = 15000
print(f"{name}은 {price:,.2%}원 입니다.")
```
