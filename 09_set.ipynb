# Set
set은 수학의 집합과 비슷한 형태로 `중복되는 데이터를 허용하지 않는다` 데이터를 저장하는 순서와 실제로 데이터가 저장되는 순서가 달라서 `인덱싱과 슬라이싱을 사용할 수 없다`는 특징이 있다.  
set은 딕셔너리 처럼 중괄호를 사용해서 데이터를 감싸고 key는 존재하지 않고 value만 존재한다.  

```python
# 빈 set 만들기
a= {} # 빈 set은 만드는 방식이 한가지인데 딕셔너리처럼 {}를 쓰기 때문에 a = {}처럼 하면 딕셔너리로 인식이 된다.

#빈 set은 반드시 생성자 함수를 사용해서 만들어야 한다.
b = set()
print(b)
print(type(b))
```

```python
a = {1,2,3,4}
print(a)
print(type(a))
```

```python
#set은 중복되는 데이터의 입력을 허용하지 않는다.
a = {1,5,1,1,1,1,1,3,5,7,7}
print(a)
type(a)
```

```python
#add() 메소드로 데이터를 set에 추가할 수 있다.
a = set()
print(a)
a.add("홍길동")
print(a)
a.add("임꺽정")
print(a)
a.add("장길산")
print(a)
a.add("일지매")
print(a)
```

```python
# remove() 또는 discard() 이 두개의 메소드로 set에서 데이터 제거 가능
```

```python
#a.remove("장길산") #remove는 set에 존재하지 않는 데이터를 삭제할려면 에러가 뜬다
a.discard("일지매") #discard는 set에 존재하지 않는 데이터를 삭제할려면 명령이 무시가 된다.
print(a)
```

# 집합 연산

```python
a = {1,2,3,4,5}
b = {3,4,5,6,7}
# 합집합
print(a.union(b))
print(a | b)

# 교집합
print(a.intersection(b))
print(a & b)

#차집합
print(a.difference(b))
print(a - b)

#대칭 차집합 => 합집합 - 교집합
print(a.symmetric_difference(b))
print(a ^ b)
```
