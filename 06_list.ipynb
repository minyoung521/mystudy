# 리스트 메소드

```python
a = [1,2,3,4,5]

#len() 함수로 리스트에 저장된 데이터의 개수를 얻어온다.
print(len(a))
# append() 메소드로 리스트의 맨 뒤에 데이터를 추가할 수 있다.
print(a)
a.append(7)
print(a)

# insert(인덱스 번호, 데이터) 메소드로 리스트의 특정 위치에 데이터 삽입할 수 있다.
a.insert(2,100)
print(a)
```

```python
#sort() 메소드로 리스트에 저장된 데이터를 정렬할 수 있다. 별도 설정 X 오름차순 reverse = True로 해서 역으로 정렬
a = [7,5,2,3,1]
a.sort()
print(a)

b = ["홍길동","임꺽정","장길산","일지매"]
b.sort()
print(b)

d = ["50","1","010","999"]
d.sort(reverse = True)
print(d)
```

sort() 메소드를 실행하는 리스트는 리스트 원본 자체가 정렬되고 sorted()함수는 함수를 실행하는 리스트는 건드리지 않고 정렬된 결과를 리턴한다.

```python
a = [1,5,3,4,2,6]

b = sorted(a,reverse=True)

print(b)
```

index() 메소드는 리스트에서 인수로 지정된 데이터가 최초로 출현하는 인덱스

```python
a = [1,2,3,1,3,1,4,1,5]
print(a.index(1))
print(a.index(2))
```

remove() 메소드로 리스트에서 인수로 지정한 첫번째 데이터만 제거
del 명령은 지정된 인덱스의 데이터를 제거하고 remove() 는 지정된 데이터를 제거 

```python
a = [1,2,3,1,3,1,4,1,5]
print(a)
a.remove(3)
print(a)
a.remove(3)
print(a)
a.pop(2)
print(a)
```

pop() 메소드는 별도의 설정이 없으면 리스트의 맨 마지막 데이터를 리턴하고 제거한다.
