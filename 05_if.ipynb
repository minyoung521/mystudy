키보드로 년도를 입력받아 윤년인가 평년인가 판단한다.
년도가 4로 나누어 떨어지고 100으로 나누어 떨어지지 않거나 400으로 나눠 떨어지면 윤년 그렇지 않으면 평년

```python
year = input("년도를 입력하십시오.")
year_int = int(year)
if (year_int % 4 == 0 and year_int % 100 != 0) or year_int % 400 == 0:
    print(f"{year}년은 윤년입니다.")
else:
    print("{}년은 평년입니다.".format(year))
```

```python
import random
```

```python
rsp = input("가위 바위 보 중에서 하나를 고르시오>> ") #0은 가위, 1은 바위, 2는 보
a = random.randint(0,2)
if a == 0:
    a = '가위'
elif a == 1:
    a = '바위'
else:
    a = '보'
    

if rsp == '가위':
    if a == '가위':
        print("컴퓨터:{} - 사용자:{} >> 비겼다...".format(a,rsp))
    elif a == '바위':
        print("컴퓨터:{} - 사용자:{} >> 이겼다!!".format(a,rsp))
    else:
        print("컴퓨터:{} - 사용자:{} >> 졌다ㅠㅠ".format(a,rsp))
elif rsp == '바위':
    if a == '가위':
        print("컴퓨터:{} - 사용자:{} >> 졌다ㅠㅠ".format(a,rsp))
    elif a == '바위':
        print("컴퓨터:{} - 사용자:{} >> 비겼다...".format(a,rsp))
    else:
        print("컴퓨터:{} - 사용자:{} >> 이겼다!!".format(a,rsp))
        
elif rsp == '보':
    if a == '가위':
        print("컴퓨터:{} - 사용자:{} >> 이겼다!".format(a,rsp))
    elif a == '바위':
        print("컴퓨터:{} - 사용자:{} >> 졌다ㅠㅠ.".format(a,rsp))
    else:
        print("컴퓨터:{} - 사용자:{} >> 비겼다...".format(a,rsp))
else:
    print("다시 입력하십시오.")
    
```
