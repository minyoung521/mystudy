# 리스트의 인덱식과 슬라이싱

```python
a = [1,2,3,4,5]
print(a)

print(a[2])
print(a[-1])

print(a[1:4])
print(a[0:3])
print(a[:3])
print(a[2:])
print(a[2:5])
```

## 리스트에 저장된 데이터의 수정 및 삭제

```python
a = [1,2,3]
print(a)
a[1] = 5
a
del a[1]
a
a.clear()
a

#del 명령어로 리스트 자체를 삭제 가능
del a
print(a)
```

# 리스트 메소드

```python
a = [1,2,3,4,5]

#len() 함수로 리스트에 저장된 데이터의 개수를 얻어온다.
print(len(a))
# append() 메소드로 리스트의 맨 뒤에 데이터를 추가할 수 있다.
print(a)
a.append(7)
print(a)

# insert(인덱스 번호, 데이터) 메소드로 리스트의 특정 위치에 데이터 삽입할 수 있다.
a.insert(2,100)
print(a)
```

```python
#sort() 메소드로 리스트에 저장된 데이터를 정렬할 수 있다. 별도 설정 X 오름차순 reverse = True로 해서 역으로 정렬
a = [7,5,2,3,1]
a.sort()
print(a)

b = ["홍길동","임꺽정","장길산","일지매"]
b.sort()
print(b)

d = ["50","1","010","999"]
d.sort(reverse = True)
print(d)
```
