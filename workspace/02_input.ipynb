# 변수(Variable)

변수는 처리할 데이터(상수, constant)를 기억하는 기억장소를 말한다.

# 변수명 작성 방법
영문자(대소문자를 구분), 숫자, 특수문자(_)

=는 대입문, 할당문, 배정문

```python
a = 10
print(a)
del a
print(a)
```

split() 메소드를 사용하면 1개의 input() 함수로 여러개의 문자열 데이터를 입력받을 수 있다.

```python
name, age = input("이름과 나이를 입력:").split(":")
print(f"{name},{age}")
```

```python
t1, t2, t3 = map(int, input("3과목 점수를 입력하세요:").split(','))
total = t1 + t2 + t3
avg = total / 3
print(f"각 과목의 점수는 {t1},{t2},{t3} 입니다")
print(f"총점 : {total}")
print(f"평균 : {avg:.2f}")
print(type(total))
print(type(avg))
```
