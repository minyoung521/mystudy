# 튜플

튜플은 리스트와 거의 비슷하지만 약간의 차이가 있다.
리스트는 데이터를 대괄호 튜플은 소괄호에다가 묶는다.
리스트는 데이터 수정 삭제가 가능하지만 튜플은 불가능하다.


```python
# 빈 튜플 만들기
a = ()
print(a)
print(type(a))

b = tuple()
print(b)
print(type(b))
```

```python
# 데이터가 1개인 리스트
a = [1]
print(a)

# 데이터가 1개인 튜플
a = (1)
print(a)
print(type(a))

#데이터가 1개인 튜플을 만들려면 ()안의 데이터 뒤에 반드시 ','를 입력해야한다.
#데이터가 1개인 튜플을 만들기 위해서 ()만 사용하면 파이썬은 ()를 튜플로 취급하지 않고 연산식의 우선순위를 의미하는 ()로 취급
a = (1,)
print(a)
print(type(a))
```

# 초기치를 가지는 튜플 만들기

```python
a = (1,2,3)
print(a)

b = ('a','b','c')
print(b)

c = ((100,'a'),(200,'b'))
print(c)
print(type(c))
```

```python
# 튜플의 인덱싱과 슬라이싱 튜플은 리스트처럼 인덱싱과 슬라이싱이 가능하다.
```

```python
a = (1,2,3,4,5)
print(a)

print(a[2])
print(a[-1])

print(a[1:4])
print(a[0:3])
print(a[:3])
print(a[2:])
print(a[2:5])
```

# 튜플에 사용되는 연산자

```python
a = (1,2,3)
b = (1,2,3,4)

print(a+b)
print(a*3)
print(len(a)/3)
```

# 튜플은 리스트와 달리 수정과 삭제가 불가능하다.

# 데이터를 ','로 구분해서 나열하면 튜플로 인식된다.

```python
a = 1,2,3,4,5
print(type(a))
print(a)
```

# 두 기억장소에 저장된 데이터 교환하기

```python
a = 3
b = 4
print(f"{a},{b}")
#모든 언어에서 공통적으로 사용할 수 있는 방법
c = a
a = b
b = c
print(f"{a},{b}")

# 파이썬에서만 사용할 수 있는 방법 - 리스트 사용
[a,b] = [b,a]
print(f"{a},{b}")
#파이썬에서만 사용할 수 있는 방법 - 튜플 사용
(a,b) = (b,a)
print(f"{a},{b}")
a,b = b,a
print(f"{a},{b}")
```
