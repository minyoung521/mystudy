len() 함수는 인수로 지정된 문자열을 구성하는 문자의 갯수를 불러온다

```python
string = "We are the Champions, My Friends! ww"
len(string)
```

count() 메소드는 문자열에서 인수로 지정한 문자열의 출현횟수를 얻어온다.

```python
string.count('W') + string.count('w')
```

index() 메소드는 문자열에서 인수로 지정한 문자열이 최초로 출현하는 인덱스를 얻어온다.

```python
string = "We are the Champions, My Friends! ww"
string.index('e')
string.index('a')

```

find() 메소드는


```python
string = "We are the Champions, My Friends! ww"
print(string.find('e'))
print(string.find('We'))
string.find('we')
```

```python
title() 메소드는 인수로 지정한 문자열의 첫 문자만 대문자로 변환한다.
```

```python
string = "eeeeeeeeeeeeeeeeeeeeeeeeee"
print(string.title())
```

strip() 메소드는 문자열 앞, 뒤에 불필요한 공백을 제거
lstrip() 앞쪽의 불필요한 공백을 제거
rstrip() 뒤쪽의 불필요한 공백을 제거

```python
string = '       samsungAXI          '
print(string)
print(string.strip())
```

replace() 메소드는 특정 문자열을 지정한 문자열로 일관 치환한다.

```python
string = "We are the champions, My friends!"
print(string.replace("e",'2'))
```

join() 메소드는 인수로 지정된 문자열 사이에 특정 문자열을 삽입해 연결한다.

```python
string = 'samsungAXI'
print('/'.join(string))
print(' '.join(string))
print('\n'.join(string))
print('!(^^)'.join(string))
```
