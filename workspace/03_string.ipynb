dedent() 함수는 여러 줄로 이루어진 문자열에서 공통된 들여쓰기를 제거할 때 사용한다.
클래스의 docstring이나 들여쓰기가 포함된 html,sql 등의 멀티라인 문자열을 깔끔하게 정렬할때 주로 사용한다.

```python
from textwrap import dedent
```

```python
memo = dedent("""
    개울가에 올챙이 한마리
    꼬물꼬물 헤엄치다
    뒷다리가 쑤욱
""")

print(memo)
```
