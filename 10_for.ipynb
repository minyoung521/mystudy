## range() 함수

range() 함수는 초기치, 최종치, 증가치에 따른 숫자 리스트를 만든다.  
range() 함수는 초기치 부터 시작해서 증가치 만큼 증가하며 최종치를 넘어가지 않는 숫자 리스트를 만든다.

`range([초기치,최종치,증가치])` []로 묶어놓은 내용은 필요에 따라서 생략할 수 있다.  
초기치와 증가치를 생략하면 초기치의 기본값 `0`이고 증가치는 기본값은 `1`이다.  
즉, range(0,10,1) 0부터 9까지 1씩 증가한다.  
&nbsp;&nbsp;&nbsp;&nbsp;range(10) 0부터 9까지 1씩 증가한다.  

```python
print(range(0,10,1))
```

range() 함수가 실행된 결과를 리스트로 확인하려면 리스트 생성자 함수 list()를 사용한다.

```python
print(list(range(0,10,1)))
print(list(range(10)))
```

```python
for i in range(1, 6):
    print(i)
```

# 주사위를 10번 굴려서 출현한 눈의 개수를 센다. => 변수 사용

```python
import random
```

```python
total = 0 
for i in range(0,10):
    a = random.randint(1,6)
    print(f"주사위에서 나온 숫자: {a}")
    total += a
print(f"주사위의 총합: {total}")
    
```

```python
number_1 = 0
number_2 = 0
number_3 = 0
number_4 = 0
number_5 = 0
number_6 = 0

for i in range(1, 11):
    dice = random.randint(1, 6)
    print(f"{i}번째 주사위 결과: {dice}") 
    
    if dice == 1:
        number_1 = number_1 + 1
    elif dice == 2:
        number_2 = number_2 + 1
    elif dice == 3:
        number_3 = number_3 + 1
    elif dice == 4:
        number_4 = number_4 + 1
    elif dice == 5:
        number_5 = number_5 + 1
    elif dice == 6:
        number_6 = number_6 + 1

print("--- 주사위 10번 던진 결과 ---")
print(f"1의 개수: {number_1}")
print(f"2의 개수: {number_2}")
print(f"3의 개수: {number_3}")
print(f"4의 개수: {number_4}")
print(f"5의 개수: {number_5}")
print(f"6의 개수: {number_6}")
```

```python
counts = [0,0,0,0,0,0]
for i in range(10):
    a = random.randint(1,6)
    print(f"주사위의 숫자:{a}")
    if a == 1:
        counts[0] += 1
    elif a == 2:
        counts[1] += 1
    elif a == 3:
        counts[2] += 1
    elif a == 4:
        counts[3] += 1
    elif a == 5:
        counts[4] += 1
    else:
        counts[5] += 1
print("--- 주사위 10번 던진 결과 ---")
print(f"눈 1의 개수 : {counts[0]}")
print(f"눈 2의 개수 : {counts[1]}")
print(f"눈 3의 개수 : {counts[2]}")
print(f"눈 4의 개수 : {counts[3]}")
print(f"눈 5의 개수 : {counts[4]}")
print(f"눈 6의 개수 : {counts[5]}")
    
```

```python
counts = [0, 0, 0, 0, 0, 0]

for i in range(10):
    a = random.randint(0, 5)
    print(f"주사위 눈 숫자: {a}")
    counts[a] += 1            

print("--- 주사위 10번 던진 결과 ---")
for i in range(6):
    print(f"눈 숫자 {i+1}: {counts[i]}번")
```

```python
lotto = []
for i in range(0,6):
    a = random.randint(1,45)
    lotto.append(a)
print("--------------최종 결과--------------")
for i in range(0,6):
    print(f"{i+1}번째 숫자 : {lotto[i]}")
```
